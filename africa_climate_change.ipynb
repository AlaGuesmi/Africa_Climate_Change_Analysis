```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objs as go
import plotly.subplots as sp

df=pd.read_csv('Africa_climate_change.csv')
df.tail()
```

```python
df.shape
```

```python
numerical_var=df.select_dtypes(include='number')
numerical_var.columns
```

```python
categorical_var=df.select_dtypes(include=object)
categorical_var.columns
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['PRCP']) #more than half of the values are null
df.isnull().sum()
```

```python
for col in ['TAVG','TMAX','TMIN']:
    df[col]=df[col].fillna(df[col].mean())
df.isnull().sum()
```

```python
sum(df.isnull().sum())
```

```python
sum(df.duplicated())
```

```python
df=df.drop_duplicates()
sum(df.duplicated())
```

```python
fig = px.line(df[(df['COUNTRY'] == "Tunisia") | (df['COUNTRY'] == "Cameroon")], x='DATE', y='TAVG', title='Average temperature fluctuations in Tunisia and Cameroon') # x is date always
fig.update_traces(line=dict(color='green'))

fig.show()
```

```python
df1=df[(df['DATE'] >= "1980-01-01") & (df["DATE"] <= "2005-12-31")]
fig_2=fig = px.line(df1, x='DATE', y='TAVG', title='Average Temperature Fluctuations (1980-2005)') # x is date always
fig_2.update_traces(line=dict(color='green'))

fig_2.show()
```

```python
fig_histograms = sp.make_subplots(rows=1, cols=2, subplot_titles=['1980-2000', '2000-2023'])
senegal_data_2000_2023 = df[(df['COUNTRY'] == 'Senegal') & (df['DATE'] >= '2000-01-01') & (df['DATE'] <= '2023-12-31')]
# Filter data for the specified date ranges
senegal_data_1980_2000 = df[(df['COUNTRY'] == 'Senegal') & (df['DATE'] >= '1980-01-01') & (df['DATE'] <= '2000-12-31')]
fig_histograms.add_trace(go.Histogram(x=senegal_data_1980_2000["TAVG"], name="1980-2000", marker_color="blue"), row=1, col=1)
fig_histograms.add_trace(go.Histogram(x=senegal_data_2000_2023["TAVG"], name="2000-2023", marker_color="red"), row=1, col=2)
fig_histograms.update_layout(title_text="Temperature Distribution in Senegal", barmode="overlay")
fig_histograms.show()
```

```python
df_average_temp = df.groupby("COUNTRY", as_index=False)["TAVG"].mean()
fig_average_temp = px.bar(df_average_temp, x="COUNTRY", y="TAVG", title="Average Temperature Per Country",
                      labels={"TAVG": "Avg Temperature (°C)", "COUNTRY": "Country"},
                      color="TAVG", color_continuous_scale="reds")
fig_average_temp.update_layout(xaxis_tickangle=-45)
fig_average_temp.show()
```
